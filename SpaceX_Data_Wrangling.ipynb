```python
import pandas as pd
import numpy as np

df = pd.read_csv('spacex_web_scraped.csv')

target_col = 'Launch outcome' 

landing_outcomes = df[target_col].value_counts()

bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])

df['Class'] = [0 if x in bad_outcomes else 1 for x in df[target_col]]

print(df['Class'].mean())

df.to_csv('dataset_part_2.csv', index=False)
df.head()
```

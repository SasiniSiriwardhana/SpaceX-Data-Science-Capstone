```python
import folium
import pandas as pd

df = pd.read_csv('dataset_part_2.csv')

launch_sites_dict = {
    'Vandenberg, SLC-4E': [34.6328, -120.6107],
    'Cape Canaveral, SLC-40': [28.5623, -80.5774],
    'Kennedy Space Center, LC-39A': [28.5732, -80.6469],
    'CCSFS, SLC-40': [28.5623, -80.5774]
}

site_map = folium.Map(location=[28.5623, -80.5774], zoom_start=4)

for index, row in df.iterrows():
    site_name = row['Launch site']

    coordinate = launch_sites_dict.get(site_name, [28.5623, -80.5774])
    
    icon_color = 'green' if row['Class'] == 1 else 'red'
    
    folium.Marker(
        location=coordinate,
        popup=f"Site: {site_name}\nStatus: {'Success' if row['Class'] == 1 else 'Failure'}",
        icon=folium.Icon(color=icon_color)
    ).add_to(site_map)

site_map
```
